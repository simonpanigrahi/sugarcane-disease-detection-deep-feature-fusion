# Notebook 01 — Data Preparation
**Sugarcane Leaf Disease Detection using Deep Feature Fusion**

This notebook covers:
- Loading and validating the dataset directory structure
- Class distribution analysis
- Stratified train/val/test split
- Saving split indices for downstream notebooks


```python
import sys, os
sys.path.append(os.path.join('..', 'src'))

import numpy as np
from utils import load_image_paths_and_labels, encode_labels, stratified_split, plot_class_distribution
```

```python
# ── Configuration ──────────────────────────────────────────────────────────
DATASET_DIR = '../data/sugarcane_dataset'   # Update this path
RANDOM_SEED = 42

# ── Load paths and labels ─────────────────────────────────────────────────
image_paths, labels, class_names = load_image_paths_and_labels(DATASET_DIR)

print(f'Total images : {len(image_paths)}')
print(f'Classes ({len(class_names)}): {class_names}')
```

```python
plot_class_distribution(labels, class_names, title='Sugarcane Disease Class Distribution')
```

```python
y, label_encoder = encode_labels(labels)

X_train, X_val, X_test, y_train, y_val, y_test = stratified_split(
    image_paths, y, test_size=0.15, val_size=0.15, random_state=RANDOM_SEED
)

print(f'Train : {len(X_train)} | Val : {len(X_val)} | Test : {len(X_test)}')

# Save splits for downstream notebooks
np.save('../data/X_train_paths.npy', X_train)
np.save('../data/X_val_paths.npy',   X_val)
np.save('../data/X_test_paths.npy',  X_test)
np.save('../data/y_train.npy', y_train)
np.save('../data/y_val.npy',   y_val)
np.save('../data/y_test.npy',  y_test)

import joblib
joblib.dump(label_encoder, '../data/label_encoder.pkl')
print('Splits saved.')
```
