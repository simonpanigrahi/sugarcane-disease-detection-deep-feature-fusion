# Notebook 02 — Data Augmentation
**Sugarcane Leaf Disease Detection using Deep Feature Fusion**

This notebook covers:
- Augmentation pipeline definition (rotation, flip, zoom, brightness)
- Visual validation of augmented samples
- TensorFlow `tf.data` pipeline construction for training


```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing.image import load_img, img_to_array
```

```python
# ── Augmentation layers ───────────────────────────────────────────────────
augmentation_pipeline = tf.keras.Sequential([
    tf.keras.layers.RandomFlip('horizontal_and_vertical'),
    tf.keras.layers.RandomRotation(0.2),
    tf.keras.layers.RandomZoom(0.15),
    tf.keras.layers.RandomBrightness(0.1),
    tf.keras.layers.RandomContrast(0.1),
], name='augmentation')

print('Augmentation pipeline defined.')
```

```python
# ── Visualise augmented samples ───────────────────────────────────────────
SAMPLE_IMAGE = '../data/sugarcane_dataset/healthy/sample.jpg'  # Update path

img = img_to_array(load_img(SAMPLE_IMAGE, target_size=(224, 224)))
img_tensor = tf.expand_dims(img, axis=0)

fig, axes = plt.subplots(2, 4, figsize=(14, 6))
axes[0, 0].imshow(img.astype('uint8'))
axes[0, 0].set_title('Original')
axes[0, 0].axis('off')

for i, ax in enumerate(axes.flat[1:]):
    aug = augmentation_pipeline(img_tensor, training=True).numpy().squeeze().astype('uint8')
    ax.imshow(aug)
    ax.set_title(f'Aug {i+1}')
    ax.axis('off')

plt.suptitle('Augmentation Examples', fontsize=13)
plt.tight_layout()
plt.show()
```
